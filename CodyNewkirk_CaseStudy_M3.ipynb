```python
# Module3_CaseStudy.ipynb
# Cody Newkirk
# SDEV220

"""
Summary:
This Python app uses object-oriented programming to model vehicle data.
It defines a base class `Vehicle` with a type attribute and a derived class `Automobile`
that includes specific attributes for cars: year, make, model, number of doors, and roof type.
The program gathers this data from the user and displays it in a readable format.
"""

# Superclass: Vehicle
class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type

# Subclass: Automobile
class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")  # Fixed as "car" for this app
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

# Collect user input
print("Enter details for your car:")

year = input("Year: ")
make = input("Make: ")
model = input("Model: ")
doors = input("Number of doors (2 or 4): ")
roof = input("Type of roof (solid or sun roof): ")

# Create Automobile object with input data
car = Automobile(year, make, model, doors, roof)

# Display formatted car information
print("\nVehicle Information:")
print(f"  Vehicle type: {car.vehicle_type}")
print(f"  Year: {car.year}")
print(f"  Make: {car.make}")
print(f"  Model: {car.model}")
print(f"  Number of doors: {car.doors}")
print(f"  Type of roof: {car.roof}")
```

```python
# 9.2-Functions.ipynb
# Cody Newkirk
# SDEV220

# Define a generator function that yields odd numbers from range(10)
def get_odds():
    for number in range(10):
        if number % 2 != 0:
            yield number

# Use a for loop to find and print the third odd number
count = 0
for odd in get_odds():
    count += 1
    if count == 3:
        print("The third odd number is:", odd)
        break
```
